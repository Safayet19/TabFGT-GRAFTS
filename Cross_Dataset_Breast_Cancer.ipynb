```python
import os
import json
import math
import random
import warnings
import logging
from typing import Dict, List, Tuple, Optional

import joblib
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    mean_squared_error,
    r2_score,
)
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.cluster import KMeans
from sklearn.linear_model import Ridge
from sklearn.feature_selection import mutual_info_classif, mutual_info_regression

import torch
import torch.nn as nn
import torch.nn.functional as F

warnings.filterwarnings("ignore")
logging.getLogger("sklearn").setLevel(logging.ERROR)

TRAIN_PATH = "/content/train.csv"
TEST_PATH = "/content/test.csv"
TARGET_COLUMN = "diagnosis"
ID_COLUMNS = []
OUTPUT_DIR = "/content/grafts_outputs"

TASK_TYPE = "auto"
RANDOM_STATE = 42
VALID_SIZE = 0.20

TOP_K_NEIGHBORS = 6
N_ROLES = 6
HIDDEN_DIM = 48
GNN_EPOCHS = 40
GNN_LR = 8e-3
ROLE_PERSISTENCE_BETA = 0.45
ROLE_STABILITY_WEIGHT = 1.20

RF_TREES = 300
N_PERTURB_TRIALS = 4
N_CONTEXT_CLUSTERS = 4
MIN_VALID_CLUSTER_SIZE = 20

STAT_GRAPH_ALPHA = 0.75
FUSED_GRAPH_ALPHA = 0.60
N_TOPOLOGY_VIEWS = 5
GRAPH_EDGE_DROP_RATE = 0.15
N_ROLE_STABILITY_VIEWS = 5
NODE_NOISE_STD = 0.03

SURVIVAL_THRESHOLD = 0.56
REDUNDANCY_THRESHOLD = 0.92
MIN_FEATURES_TO_KEEP = 5
MAX_FEATURES_TO_KEEP = 50
MIN_ROLE_DIVERSE_SEEDS = 1

RANKING_MARGIN = 0.03
MAX_RANK_PAIRS = 1024

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

def seed_everything(seed: int = 42) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def print_rule(title: str) -> None:
    return

def ensure_dir(path: str) -> None:
    os.makedirs(path, exist_ok=True)

def safe_minmax(x: np.ndarray) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    if x.size == 0:
        return x.astype(float)
    finite_mask = np.isfinite(x)
    if not finite_mask.any():
        return np.zeros_like(x, dtype=float)
    fill_val = np.nanmedian(x[finite_mask]) if finite_mask.any() else 0.0
    x = np.where(finite_mask, x, fill_val)
    x_min = np.min(x)
    x_max = np.max(x)
    if np.isclose(x_min, x_max):
        return np.full_like(x, 0.5, dtype=float)
    return (x - x_min) / (x_max - x_min + 1e-12)

def sigmoid_np(x: np.ndarray) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    return 1.0 / (1.0 + np.exp(-x))

def read_table(path: Optional[str]) -> Optional[pd.DataFrame]:
    if path is None:
        return None
    ext = os.path.splitext(path)[1].lower()
    if ext == ".csv":
        return pd.read_csv(path)
    if ext in {".xlsx", ".xls"}:
        return pd.read_excel(path)
    if ext == ".parquet":
        return pd.read_parquet(path)
    raise ValueError(f"Unsupported file type: {ext}. Use CSV, Excel, or Parquet.")

def infer_task_type(y: pd.Series) -> str:
    if str(y.dtype) in {"object", "category", "bool", "string"}:
        return "classification"
    if pd.api.types.is_integer_dtype(y) and y.nunique(dropna=True) <= max(20, int(0.05 * len(y))):
        return "classification"
    if pd.api.types.is_float_dtype(y) and y.nunique(dropna=True) <= max(15, int(0.03 * len(y))):
        return "classification"
    return "regression"

def score_model(model, X: pd.DataFrame, y: np.ndarray, task_type: str) -> float:
    pred = model.predict(X)
    if task_type == "classification":
        if len(np.unique(y)) == 1:
            return accuracy_score(y, pred)
        return f1_score(y, pred, average="macro")
    rmse = math.sqrt(mean_squared_error(y, pred))
    return -rmse

def evaluate_subset(
    task_type: str,
    X_train: pd.DataFrame,
    y_train: np.ndarray,
    X_valid: pd.DataFrame,
    y_valid: np.ndarray,
    random_state: int,
    n_trees: int,
) -> Dict[str, float]:
    model = build_base_model(task_type, random_state, n_trees)
    model.fit(X_train, y_train)
    pred = model.predict(X_valid)

    if task_type == "classification":
        out = {
            "accuracy": float(accuracy_score(y_valid, pred)),
            "macro_f1": float(f1_score(y_valid, pred, average="macro")),
        }
        return out

    rmse = math.sqrt(mean_squared_error(y_valid, pred))
    out = {
        "rmse": float(rmse),
        "r2": float(r2_score(y_valid, pred)),
    }
    return out

def unique_preserve_order(items: List[str]) -> List[str]:
    seen = set()
    out = []
    for item in items:
        if item not in seen:
            seen.add(item)
            out.append(item)
    return out

class LeakageSafeTabularEncoder:
    def __init__(self):
        self.columns_: List[str] = []
        self.numeric_cols_: List[str] = []
        self.categorical_cols_: List[str] = []
        self.num_median_: Dict[str, float] = {}
        self.num_mean_: Dict[str, float] = {}
        self.num_std_: Dict[str, float] = {}
        self.cat_freq_: Dict[str, Dict[str, float]] = {}

    def fit(self, X: pd.DataFrame):
        X = X.copy()
        self.columns_ = X.columns.tolist()
        self.numeric_cols_ = X.select_dtypes(include=[np.number, "bool"]).columns.tolist()
        self.categorical_cols_ = [c for c in self.columns_ if c not in self.numeric_cols_]

        for col in self.numeric_cols_:
            s = pd.to_numeric(X[col], errors="coerce")
            median = float(s.median()) if s.notna().any() else 0.0
            s = s.fillna(median)
            mean = float(s.mean())
            std = float(s.std())
            if not np.isfinite(std) or std < 1e-12:
                std = 1.0
            self.num_median_[col] = median
            self.num_mean_[col] = mean
            self.num_std_[col] = std

        for col in self.categorical_cols_:
            s = X[col].astype("string").fillna("__MISSING__")
            freq = s.value_counts(normalize=True, dropna=False).to_dict()
            self.cat_freq_[col] = {str(k): float(v) for k, v in freq.items()}

        return self

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        if not self.columns_:
            raise RuntimeError("Encoder is not fitted yet.")

        X = X.copy()
        missing_cols = [c for c in self.columns_ if c not in X.columns]
        if missing_cols:
            raise ValueError(f"Input is missing columns required by encoder: {missing_cols}")

        out = pd.DataFrame(index=X.index)

        for col in self.numeric_cols_:
            s = pd.to_numeric(X[col], errors="coerce").fillna(self.num_median_[col]).astype(float)
            out[col] = (s - self.num_mean_[col]) / self.num_std_[col]

        for col in self.categorical_cols_:
            s = X[col].astype("string").fillna("__MISSING__")
            out[col] = s.map(self.cat_freq_[col]).fillna(0.0).astype(float)

        return out[self.columns_]

    def fit_transform(self, X: pd.DataFrame) -> pd.DataFrame:
        return self.fit(X).transform(X)

    def get_state(self) -> Dict:
        return {
            "columns_": self.columns_,
            "numeric_cols_": self.numeric_cols_,
            "categorical_cols_": self.categorical_cols_,
            "num_median_": self.num_median_,
            "num_mean_": self.num_mean_,
            "num_std_": self.num_std_,
            "cat_freq_": self.cat_freq_,
        }

def build_base_model(task_type: str, random_state: int, n_trees: int):
    if task_type == "classification":
        return RandomForestClassifier(
            n_estimators=n_trees,
            random_state=random_state,
            n_jobs=-1,
            class_weight="balanced_subsample",
            min_samples_leaf=2,
        )
    return RandomForestRegressor(
        n_estimators=n_trees,
        random_state=random_state,
        n_jobs=-1,
        min_samples_leaf=2,
    )

def normalize_adjacency(A: np.ndarray) -> np.ndarray:
    A_tilde = A + np.eye(A.shape[0], dtype=float)
    d = A_tilde.sum(axis=1)
    d_inv_sqrt = np.diag(1.0 / np.sqrt(d + 1e-12))
    return d_inv_sqrt @ A_tilde @ d_inv_sqrt

def topk_symmetric_graph(sim: np.ndarray, top_k: int) -> np.ndarray:
    sim = np.asarray(sim, dtype=float).copy()
    np.fill_diagonal(sim, 0.0)

    A = np.zeros_like(sim, dtype=float)
    n = sim.shape[0]

    for i in range(n):
        row = sim[i]
        if np.allclose(row, 0.0):
            continue
        idx = np.argsort(row)[-top_k:]
        for j in idx:
            if row[j] > 0:
                A[i, j] = float(row[j])

    A = np.maximum(A, A.T)
    if A.max() > 0:
        A = A / (A.max() + 1e-12)
    return A

def build_statistical_graph(
    X_train_raw: pd.DataFrame,
    X_train_enc: pd.DataFrame,
    top_k: int,
    alpha: float = STAT_GRAPH_ALPHA,
) -> Tuple[pd.DataFrame, pd.DataFrame, np.ndarray, np.ndarray]:

    corr = X_train_enc.corr(method="spearman").fillna(0.0).abs()
    np.fill_diagonal(corr.values, 0.0)

    miss_ind = X_train_raw.isna().astype(float)
    if miss_ind.shape[1] > 1:
        miss_corr = miss_ind.corr(method="pearson").fillna(0.0).abs()
    else:
        miss_corr = pd.DataFrame(np.zeros((1, 1)), index=X_train_raw.columns, columns=X_train_raw.columns)
    np.fill_diagonal(miss_corr.values, 0.0)

    stat_sim = alpha * corr.values + (1.0 - alpha) * miss_corr.values
    A_stat = topk_symmetric_graph(stat_sim, top_k=top_k)
    A_stat_norm = normalize_adjacency(A_stat)

    return corr, miss_corr, A_stat, A_stat_norm

def cosine_similarity_matrix(X: np.ndarray) -> np.ndarray:
    X = np.asarray(X, dtype=float)
    norms = np.linalg.norm(X, axis=1, keepdims=True)
    Xn = X / np.clip(norms, 1e-12, None)
    sim = Xn @ Xn.T
    sim = np.clip(sim, 0.0, 1.0)
    np.fill_diagonal(sim, 0.0)
    return sim

def build_response_graph(
    response_profile_df: pd.DataFrame,
    top_k: int,
) -> Tuple[np.ndarray, np.ndarray]:
    sim = cosine_similarity_matrix(response_profile_df.values)
    A_resp = topk_symmetric_graph(sim, top_k=top_k)
    A_resp_norm = normalize_adjacency(A_resp)
    return A_resp, A_resp_norm

def fuse_graphs(
    A_stat: np.ndarray,
    A_resp: np.ndarray,
    alpha: float = FUSED_GRAPH_ALPHA,
) -> Tuple[np.ndarray, np.ndarray]:
    A_fused = alpha * A_stat + (1.0 - alpha) * A_resp
    np.fill_diagonal(A_fused, 0.0)
    if A_fused.max() > 0:
        A_fused = A_fused / (A_fused.max() + 1e-12)
    A_fused_norm = normalize_adjacency(A_fused)
    return A_fused, A_fused_norm

def augment_graph_and_nodes(
    x: torch.Tensor,
    adj: torch.Tensor,
    edge_drop_rate: float = GRAPH_EDGE_DROP_RATE,
    node_noise_std: float = NODE_NOISE_STD,
) -> Tuple[torch.Tensor, torch.Tensor]:

    x_aug = x + torch.randn_like(x) * node_noise_std

    A = adj.detach().cpu().numpy().copy()
    n = A.shape[0]
    upper = np.triu_indices(n, k=1)
    keep_mask = (np.random.rand(len(upper[0])) > edge_drop_rate).astype(float)

    A2 = A.copy()
    A2[upper] = A2[upper] * keep_mask
    A2[(upper[1], upper[0])] = A2[upper]
    A2 = normalize_adjacency(np.clip(A2, 0.0, None))
    adj_aug = torch.tensor(A2, dtype=torch.float32, device=x.device)

    return x_aug, adj_aug

def connected_pairs(binary_adj: np.ndarray) -> int:
    n = binary_adj.shape[0]
    visited = np.zeros(n, dtype=bool)
    total_pairs = 0

    for start in range(n):
        if visited[start]:
            continue
        stack = [start]
        visited[start] = True
        size = 0

        while stack:
            node = stack.pop()
            size += 1
            neighbors = np.where(binary_adj[node] > 0)[0]
            for nb in neighbors:
                if not visited[nb]:
                    visited[nb] = True
                    stack.append(nb)

        total_pairs += size * (size - 1) // 2

    return total_pairs

def structural_survival_scores(A: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    degree = A.sum(axis=1)
    degree_norm = safe_minmax(degree)

    binary_adj = (A > 0).astype(int)
    base_pairs = max(connected_pairs(binary_adj), 1)

    connectivity_damage = np.zeros(A.shape[0], dtype=float)
    for i in range(A.shape[0]):
        reduced = binary_adj.copy()
        reduced[i, :] = 0
        reduced[:, i] = 0
        pairs_after = connected_pairs(reduced)
        connectivity_damage[i] = max(base_pairs - pairs_after, 0) / base_pairs

    connectivity_damage_norm = safe_minmax(connectivity_damage)
    structural = 0.50 * degree_norm + 0.50 * connectivity_damage_norm
    return structural, degree_norm, connectivity_damage_norm

def compute_topology_persistence(
    A_fused: np.ndarray,
    n_views: int = N_TOPOLOGY_VIEWS,
    drop_rate: float = GRAPH_EDGE_DROP_RATE,
) -> Tuple[np.ndarray, pd.DataFrame]:
    base_structural, _, _ = structural_survival_scores(A_fused)
    views = [base_structural]

    n = A_fused.shape[0]
    upper = np.triu_indices(n, k=1)

    for v in range(n_views):
        rng = np.random.default_rng(RANDOM_STATE + 3000 + v)
        A_view = A_fused.copy()

        drop_mask = (rng.random(len(upper[0])) > drop_rate).astype(float)
        A_view[upper] = A_view[upper] * drop_mask
        A_view[(upper[1], upper[0])] = A_view[upper]

        structural_v, _, _ = structural_survival_scores(A_view)
        views.append(structural_v)

    M = np.vstack(views).T
    mean_val = M.mean(axis=1)
    std_val = M.std(axis=1)
    stability = np.clip(1.0 - (std_val / (mean_val + 1e-12)), 0.0, 1.0)

    topology_persistence = np.clip(0.60 * safe_minmax(mean_val) + 0.40 * stability, 0.0, 1.0)

    detail_df = pd.DataFrame(M, columns=[f"topology_view_{i}" for i in range(M.shape[1])])
    detail_df["topology_mean"] = mean_val
    detail_df["topology_std"] = std_val
    detail_df["topology_stability"] = stability
    detail_df["topology_persistence"] = topology_persistence

    return topology_persistence, detail_df

def feature_target_abs_corr(x: np.ndarray, y: np.ndarray) -> float:
    if np.nanstd(x) < 1e-12 or np.nanstd(y) < 1e-12:
        return 0.0
    try:
        v = abs(pd.Series(x).corr(pd.Series(y), method="spearman"))
        return 0.0 if pd.isna(v) else float(v)
    except Exception:
        return 0.0

def class_centroid_separation(X: np.ndarray, y: np.ndarray) -> float:
    classes = np.unique(y)
    if len(classes) <= 1:
        return 0.0

    centroids = []
    for c in classes:
        centroids.append(X[y == c].mean(axis=0))
    centroids = np.vstack(centroids)

    dists = []
    for i in range(len(centroids)):
        for j in range(i + 1, len(centroids)):
            dists.append(np.linalg.norm(centroids[i] - centroids[j]))
    return float(np.mean(dists)) if dists else 0.0

def generate_perturbed_valid(
    X_valid: pd.DataFrame,
    X_train: pd.DataFrame,
    seed: int,
    trial_id: int,
) -> pd.DataFrame:
    rng = np.random.default_rng(seed + trial_id)
    Xp = X_valid.copy()
    mode = trial_id % 3

    if mode == 0:

        mask = rng.random(Xp.shape) < 0.10
        fill_vals = np.tile(X_train.median().values, (len(Xp), 1))
        fill_df = pd.DataFrame(fill_vals, columns=Xp.columns, index=Xp.index)
        Xp = Xp.mask(mask, fill_df)

    elif mode == 1:

        scales = X_train.std().replace(0, 1.0).values * 0.10
        noise = rng.normal(0, 1, size=Xp.shape) * scales
        Xp = Xp + noise

    else:

        n_cols = max(1, int(0.25 * Xp.shape[1]))
        chosen_cols = rng.choice(Xp.columns.to_numpy(), size=n_cols, replace=False)
        for col in chosen_cols:
            row_idx = Xp.sample(frac=0.15, random_state=int(rng.integers(1, 10**9))).index
            Xp.loc[row_idx, col] = rng.permutation(Xp.loc[row_idx, col].values)

    return Xp

def redundancy_survival_scores(
    X_train_enc: pd.DataFrame,
    corr_matrix: pd.DataFrame,
    top_k: int,
) -> Tuple[np.ndarray, np.ndarray]:
    features = X_train_enc.columns.tolist()
    redundancy_raw = np.zeros(len(features), dtype=float)

    for i, feat in enumerate(features):
        neighbor_scores = corr_matrix.loc[feat].sort_values(ascending=False)
        neighbors = [c for c in neighbor_scores.index.tolist() if c != feat and neighbor_scores[c] > 0][:top_k]

        if not neighbors:
            redundancy_raw[i] = 0.0
            continue

        y_feat = X_train_enc[feat].values
        if np.std(y_feat) < 1e-12:
            redundancy_raw[i] = 1.0
            continue

        X_neighbors = X_train_enc[neighbors].values
        ridge = Ridge(alpha=1.0)
        ridge.fit(X_neighbors, y_feat)
        r2 = max(0.0, float(ridge.score(X_neighbors, y_feat)))
        max_corr = float(corr_matrix.loc[feat, neighbors].max()) if neighbors else 0.0

        redundancy_raw[i] = np.clip(0.70 * r2 + 0.30 * max_corr, 0.0, 1.0)

    nonredundancy = 1.0 - redundancy_raw
    return redundancy_raw, nonredundancy

def compute_counterfactual_substitute_margin(
    X_train_enc: pd.DataFrame,
    X_valid_enc: pd.DataFrame,
    y_valid: np.ndarray,
    model,
    base_score: float,
    corr_matrix: pd.DataFrame,
    top_k: int,
    task_type: str,
) -> np.ndarray:
    features = X_train_enc.columns.tolist()
    raw_margin = np.zeros(len(features), dtype=float)

    for i, feat in enumerate(features):
        rng = np.random.default_rng(RANDOM_STATE + 70_000 + i)

        X_perm = X_valid_enc.copy()
        X_perm[feat] = rng.permutation(X_perm[feat].values)
        perm_score = score_model(model, X_perm, y_valid, task_type)
        drop_without_f = max(base_score - perm_score, 0.0)

        neighbor_scores = corr_matrix.loc[feat].sort_values(ascending=False)
        neighbors = [c for c in neighbor_scores.index.tolist() if c != feat and neighbor_scores[c] > 0][:top_k]

        if not neighbors or np.std(X_train_enc[feat].values) < 1e-12:
            raw_margin[i] = drop_without_f
            continue

        ridge = Ridge(alpha=1.0)
        ridge.fit(X_train_enc[neighbors].values, X_train_enc[feat].values)
        reconstructed_feat = ridge.predict(X_valid_enc[neighbors].values)

        X_recovered = X_perm.copy()
        X_recovered[feat] = reconstructed_feat
        recovered_score = score_model(model, X_recovered, y_valid, task_type)

        recovery = max(recovered_score - perm_score, 0.0)
        raw_margin[i] = max(drop_without_f - recovery, 0.0)

    return safe_minmax(raw_margin)

def compute_observed_components_and_profiles(
    X_train_raw: pd.DataFrame,
    X_train_enc: pd.DataFrame,
    y_train: np.ndarray,
    X_valid_enc: pd.DataFrame,
    y_valid: np.ndarray,
    corr_matrix: pd.DataFrame,
    A_fused: np.ndarray,
    task_type: str,
) -> Tuple[pd.DataFrame, pd.DataFrame, object]:
    features = X_train_enc.columns.tolist()

    model = build_base_model(task_type, RANDOM_STATE, RF_TREES)
    model.fit(X_train_enc, y_train)
    base_score = score_model(model, X_valid_enc, y_valid, task_type)

    structural, degree_norm, connectivity_damage_norm = structural_survival_scores(A_fused)
    topology_persistence, topology_detail_df = compute_topology_persistence(A_fused)
    redundancy_raw, nonredundancy = redundancy_survival_scores(X_train_enc, corr_matrix, TOP_K_NEIGHBORS)

    raw_semantic = np.zeros(len(features), dtype=float)
    raw_contextual = np.zeros(len(features), dtype=float)
    raw_perturbation = np.zeros(len(features), dtype=float)

    if task_type == "classification":
        base_sep = class_centroid_separation(X_valid_enc.values, y_valid)
        base_feature_corr = None
    else:
        base_feature_corr = np.array(
            [feature_target_abs_corr(X_valid_enc[f].values, y_valid) for f in features],
            dtype=float,
        )
        base_sep = None

    n_clusters = max(2, min(N_CONTEXT_CLUSTERS, len(X_train_enc)))
    kmeans = KMeans(n_clusters=n_clusters, random_state=RANDOM_STATE, n_init=10)
    kmeans.fit(X_train_enc)
    valid_clusters = kmeans.predict(X_valid_enc)

    cluster_base_scores = {}
    valid_cluster_ids = sorted(np.unique(valid_clusters).tolist())
    for cl in valid_cluster_ids:
        mask = valid_clusters == cl
        if mask.sum() >= max(2, MIN_VALID_CLUSTER_SIZE):
            cluster_base_scores[cl] = score_model(model, X_valid_enc.loc[mask], y_valid[mask], task_type)

    perturb_trial_store = {feat: [] for feat in features}
    context_cluster_store = {feat: {cl: 0.0 for cl in valid_cluster_ids} for feat in features}

    for i, feat in enumerate(features):
        rng = np.random.default_rng(RANDOM_STATE + i)

        Xp = X_valid_enc.copy()
        Xp[feat] = rng.permutation(Xp[feat].values)

        perm_score = score_model(model, Xp, y_valid, task_type)
        predictive_drop = max(base_score - perm_score, 0.0)

        if task_type == "classification":
            perm_sep = class_centroid_separation(Xp.values, y_valid)
            coherence_drop = max(base_sep - perm_sep, 0.0)
        else:
            perm_corr = feature_target_abs_corr(Xp[feat].values, y_valid)
            coherence_drop = max(base_feature_corr[i] - perm_corr, 0.0)

        raw_semantic[i] = 0.70 * predictive_drop + 0.30 * coherence_drop

        cluster_impacts = []
        for cl, base_cl_score in cluster_base_scores.items():
            mask = valid_clusters == cl
            Xpc = X_valid_enc.loc[mask].copy()
            rng_cl = np.random.default_rng(RANDOM_STATE + 10_000 + i + int(cl))
            Xpc[feat] = rng_cl.permutation(Xpc[feat].values)

            perm_cl_score = score_model(model, Xpc, y_valid[mask], task_type)
            impact = max(base_cl_score - perm_cl_score, 0.0)

            cluster_impacts.append(impact)
            context_cluster_store[feat][cl] = float(impact)

        if cluster_impacts:
            mean_imp = float(np.mean(cluster_impacts))
            stability = float(np.clip(1.0 - (np.std(cluster_impacts) / (mean_imp + 1e-12)), 0.0, 1.0))
            raw_contextual[i] = mean_imp * (0.50 + 0.50 * stability)
        else:
            raw_contextual[i] = 0.0

    for trial in range(N_PERTURB_TRIALS):
        X_pert = generate_perturbed_valid(X_valid_enc, X_train_enc, RANDOM_STATE + 50_000, trial)
        base_pert_score = score_model(model, X_pert, y_valid, task_type)

        for i, feat in enumerate(features):
            Xpf = X_pert.copy()
            rng = np.random.default_rng(RANDOM_STATE + 20_000 + trial * 1000 + i)
            Xpf[feat] = rng.permutation(Xpf[feat].values)
            perm_pert_score = score_model(model, Xpf, y_valid, task_type)
            trial_importance = max(base_pert_score - perm_pert_score, 0.0)
            perturb_trial_store[feat].append(float(trial_importance))

    for i, feat in enumerate(features):
        vals = np.array(perturb_trial_store[feat], dtype=float)
        mean_val = float(np.mean(vals))
        stability = float(np.clip(1.0 - (np.std(vals) / (mean_val + 1e-12)), 0.0, 1.0))
        raw_perturbation[i] = mean_val * (0.50 + 0.50 * stability)

    semantic = safe_minmax(raw_semantic)
    contextual = safe_minmax(raw_contextual)
    perturbation = safe_minmax(raw_perturbation)

    substitute_margin = compute_counterfactual_substitute_margin(
        X_train_enc=X_train_enc,
        X_valid_enc=X_valid_enc,
        y_valid=y_valid,
        model=model,
        base_score=base_score,
        corr_matrix=corr_matrix,
        top_k=TOP_K_NEIGHBORS,
        task_type=task_type,
    )

    component_df = pd.DataFrame({
        "feature": features,
        "structural": structural,
        "semantic": semantic,
        "perturbation": perturbation,
        "contextual": contextual,
        "non_redundancy": nonredundancy,
        "topology_persistence": topology_persistence,
        "substitute_margin": substitute_margin,
        "degree_norm": degree_norm,
        "connectivity_damage_norm": connectivity_damage_norm,
        "redundancy_raw": redundancy_raw,
    })

    response_profile_rows = []
    for i, feat in enumerate(features):
        cluster_profile = [context_cluster_store[feat][cl] for cl in valid_cluster_ids]
        pert_profile = perturb_trial_store[feat]

        row = {
            "feature": feat,
            "semantic_raw": raw_semantic[i],
            "contextual_raw": raw_contextual[i],
            "perturbation_raw": raw_perturbation[i],
            "redundancy_raw": redundancy_raw[i],
            "non_redundancy": nonredundancy[i],
            "topology_persistence": topology_persistence[i],
            "substitute_margin": substitute_margin[i],
            "cluster_mean": float(np.mean(cluster_profile)) if len(cluster_profile) > 0 else 0.0,
            "cluster_std": float(np.std(cluster_profile)) if len(cluster_profile) > 0 else 0.0,
            "pert_mean": float(np.mean(pert_profile)) if len(pert_profile) > 0 else 0.0,
            "pert_std": float(np.std(pert_profile)) if len(pert_profile) > 0 else 0.0,
        }
        for j, v in enumerate(cluster_profile):
            row[f"context_cluster_{j}"] = float(v)
        for j, v in enumerate(pert_profile):
            row[f"pert_trial_{j}"] = float(v)
        response_profile_rows.append(row)

    response_profile_df = pd.DataFrame(response_profile_rows).set_index("feature")
    response_profile_df = response_profile_df.apply(lambda col: safe_minmax(col.values), axis=0, result_type="broadcast")
    response_profile_df = response_profile_df.fillna(0.0)

    return component_df, response_profile_df, model

def build_node_feature_matrix(
    X_train_raw: pd.DataFrame,
    X_train_enc: pd.DataFrame,
    y_train: np.ndarray,
    corr_matrix: pd.DataFrame,
    A_stat: np.ndarray,
    A_resp: np.ndarray,
    A_fused: np.ndarray,
    component_df: pd.DataFrame,
    task_type: str,
) -> pd.DataFrame:
    missing_rate = X_train_raw.isna().mean().values
    unique_ratio = np.array([
        X_train_raw[c].nunique(dropna=True) / max(1, len(X_train_raw))
        for c in X_train_raw.columns
    ], dtype=float)
    encoded_std = X_train_enc.std().fillna(0.0).values
    graph_degree_stat = A_stat.sum(axis=1)
    graph_degree_resp = A_resp.sum(axis=1)
    graph_degree_fused = A_fused.sum(axis=1)
    mean_neighbor_corr = corr_matrix.mean(axis=1).values

    try:
        if task_type == "classification":
            mi = mutual_info_classif(X_train_enc, y_train, random_state=RANDOM_STATE)
        else:
            mi = mutual_info_regression(X_train_enc, y_train, random_state=RANDOM_STATE)
    except Exception:
        mi = np.zeros(X_train_enc.shape[1], dtype=float)

    node_df = pd.DataFrame({
        "mi": safe_minmax(np.asarray(mi, dtype=float)),
        "missing_rate": safe_minmax(missing_rate),
        "unique_ratio": safe_minmax(unique_ratio),
        "encoded_std": safe_minmax(encoded_std),
        "graph_degree_stat": safe_minmax(graph_degree_stat),
        "graph_degree_resp": safe_minmax(graph_degree_resp),
        "graph_degree_fused": safe_minmax(graph_degree_fused),
        "mean_neighbor_corr": safe_minmax(mean_neighbor_corr),
        "structural_seed": component_df["structural"].values,
        "topology_seed": component_df["topology_persistence"].values,
        "substitute_seed": component_df["substitute_margin"].values,
    }, index=X_train_enc.columns)

    return node_df.fillna(0.0)

class GRAFTSGNN(nn.Module):
    def __init__(self, in_dim: int, hidden_dim: int, n_roles: int, n_components: int, beta: float):
        super().__init__()
        self.lin_in = nn.Linear(in_dim, hidden_dim)
        self.lin_self1 = nn.Linear(hidden_dim, hidden_dim)
        self.lin_msg1 = nn.Linear(hidden_dim, hidden_dim)
        self.lin_self2 = nn.Linear(hidden_dim, hidden_dim)
        self.lin_msg2 = nn.Linear(hidden_dim, hidden_dim)

        self.comp_head = nn.Linear(hidden_dim, n_components)
        self.role_head = nn.Linear(hidden_dim, n_roles)

        self.role_embeddings = nn.Parameter(torch.randn(n_roles, hidden_dim) * 0.02)
        self.role_component_weights = nn.Parameter(torch.randn(n_roles, n_components) * 0.02)

        self.gamma = nn.Parameter(torch.tensor(4.0))
        self.eta = nn.Parameter(torch.tensor(1.5))
        self.mu = nn.Parameter(torch.tensor(1.5))
        self.bias = nn.Parameter(torch.tensor(-2.0))

        self.beta = beta

    def forward(
        self,
        x: torch.Tensor,
        adj: torch.Tensor,
        component_evidence: Optional[torch.Tensor] = None,
        extra_evidence: Optional[torch.Tensor] = None,
    ):
        h0 = F.gelu(self.lin_in(x))
        h1 = F.gelu(self.lin_self1(h0) + self.lin_msg1(adj @ h0))
        h2 = F.gelu(self.lin_self2(h1) + self.lin_msg2(adj @ h1))

        role_logits = self.role_head(h2)
        role_probs = torch.softmax(role_logits, dim=1)

        role_context = role_probs @ self.role_embeddings
        latent = F.gelu(h2 + role_context)

        component_pred = torch.sigmoid(self.comp_head(latent))
        evidence = component_pred if component_evidence is None else component_evidence

        role_weight_matrix = torch.softmax(self.role_component_weights, dim=1)
        omega = role_probs @ role_weight_matrix

        role_entropy = -(role_probs * torch.log(role_probs + 1e-12)).sum(dim=1) / math.log(role_probs.shape[1] + 1e-12)
        role_persistence = 1.0 - role_entropy

        core_score = (omega * evidence).sum(dim=1)

        if extra_evidence is None:
            topology_term = torch.zeros_like(core_score)
            substitute_term = torch.zeros_like(core_score)
        else:
            topology_term = extra_evidence[:, 0]
            substitute_term = extra_evidence[:, 1]

        survival_logit = (
            self.gamma * core_score
            + self.beta * role_persistence
            + self.eta * topology_term
            + self.mu * substitute_term
            + self.bias
        )
        survival = torch.sigmoid(survival_logit)

        return {
            "latent": latent,
            "component_pred": component_pred,
            "role_probs": role_probs,
            "role_persistence": role_persistence,
            "role_weight_matrix": role_weight_matrix,
            "omega": omega,
            "core_score": core_score,
            "survival_logit": survival_logit,
            "survival": survival,
        }

def pairwise_ranking_loss(
    scores: torch.Tensor,
    target: torch.Tensor,
    margin: float = RANKING_MARGIN,
    max_pairs: int = MAX_RANK_PAIRS,
) -> torch.Tensor:
    n = scores.shape[0]
    if n < 2:
        return torch.tensor(0.0, device=scores.device)

    idx1 = torch.randint(0, n, (max_pairs,), device=scores.device)
    idx2 = torch.randint(0, n, (max_pairs,), device=scores.device)

    t1 = target[idx1]
    t2 = target[idx2]
    s1 = scores[idx1]
    s2 = scores[idx2]

    valid = (idx1 != idx2) & (torch.abs(t1 - t2) > 1e-6)
    if valid.sum() == 0:
        return torch.tensor(0.0, device=scores.device)

    idx1 = idx1[valid]
    idx2 = idx2[valid]
    t1 = t1[valid]
    t2 = t2[valid]
    s1 = s1[valid]
    s2 = s2[valid]

    should_be_higher = t1 > t2
    s_pos = torch.where(should_be_higher, s1, s2)
    s_neg = torch.where(should_be_higher, s2, s1)

    loss = F.relu(margin - (s_pos - s_neg))
    return loss.mean()

def rowwise_js_divergence(p: torch.Tensor, q: torch.Tensor) -> torch.Tensor:
    p = torch.clamp(p, 1e-12, 1.0)
    q = torch.clamp(q, 1e-12, 1.0)
    m = 0.5 * (p + q)
    kl_pm = (p * (torch.log(p) - torch.log(m))).sum(dim=1)
    kl_qm = (q * (torch.log(q) - torch.log(m))).sum(dim=1)
    js = 0.5 * (kl_pm + kl_qm)
    return js

def train_grafts_gnn(
    node_features: pd.DataFrame,
    A_fused_norm: np.ndarray,
    component_df: pd.DataFrame,
):
    x = torch.tensor(node_features.values, dtype=torch.float32, device=DEVICE)
    adj = torch.tensor(A_fused_norm, dtype=torch.float32, device=DEVICE)

    core_comp_target = torch.tensor(
        component_df[["structural", "semantic", "perturbation", "contextual", "non_redundancy"]].values,
        dtype=torch.float32,
        device=DEVICE,
    )

    extra_target = torch.tensor(
        component_df[["topology_persistence", "substitute_margin"]].values,
        dtype=torch.float32,
        device=DEVICE,
    )

    observed_survival_target_np = (
        0.18 * component_df["structural"].values
        + 0.22 * component_df["semantic"].values
        + 0.18 * component_df["perturbation"].values
        + 0.14 * component_df["contextual"].values
        + 0.10 * component_df["non_redundancy"].values
        + 0.09 * component_df["topology_persistence"].values
        + 0.09 * component_df["substitute_margin"].values
    )
    observed_survival_target_np = safe_minmax(observed_survival_target_np)
    survival_target = torch.tensor(observed_survival_target_np, dtype=torch.float32, device=DEVICE)

    model = GRAFTSGNN(
        in_dim=node_features.shape[1],
        hidden_dim=HIDDEN_DIM,
        n_roles=N_ROLES,
        n_components=5,
        beta=ROLE_PERSISTENCE_BETA,
    ).to(DEVICE)

    optimizer = torch.optim.AdamW(model.parameters(), lr=GNN_LR, weight_decay=1e-4)
    L = torch.diag(adj.sum(dim=1)) - adj

    model.train()
    for epoch in range(GNN_EPOCHS):
        optimizer.zero_grad()

        out = model(
            x=x,
            adj=adj,
            component_evidence=core_comp_target,
            extra_evidence=extra_target,
        )

        comp_loss = F.mse_loss(out["component_pred"], core_comp_target)
        surv_loss = F.mse_loss(out["survival"], survival_target)
        rank_loss = pairwise_ranking_loss(out["survival"], survival_target)

        smooth_loss = torch.trace(out["latent"].T @ L @ out["latent"]) / (
            out["latent"].shape[0] * out["latent"].shape[1] + 1e-12
        )

        balance_loss = ((out["role_probs"].mean(dim=0) - (1.0 / N_ROLES)) ** 2).mean()

        x_aug, adj_aug = augment_graph_and_nodes(x, adj)
        out_aug = model(
            x=x_aug,
            adj=adj_aug,
            component_evidence=core_comp_target,
            extra_evidence=extra_target,
        )
        consistency_loss = rowwise_js_divergence(out["role_probs"], out_aug["role_probs"]).mean()

        loss = (
            comp_loss
            + 0.70 * surv_loss
            + 0.20 * rank_loss
            + 0.01 * smooth_loss
            + 0.01 * balance_loss
            + 0.05 * consistency_loss
        )
        loss.backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        out = model(
            x=x,
            adj=adj,
            component_evidence=core_comp_target,
            extra_evidence=extra_target,
        )

    return model, out

def compute_role_stability(
    model: nn.Module,
    node_features: pd.DataFrame,
    A_fused_norm: np.ndarray,
    n_views: int = N_ROLE_STABILITY_VIEWS,
) -> np.ndarray:
    x = torch.tensor(node_features.values, dtype=torch.float32, device=DEVICE)
    adj = torch.tensor(A_fused_norm, dtype=torch.float32, device=DEVICE)

    model.eval()
    with torch.no_grad():
        base = model(x=x, adj=adj)
        base_probs = base["role_probs"]

        js_list = []
        for v in range(n_views):
            x_aug, adj_aug = augment_graph_and_nodes(x, adj)
            out_v = model(x=x_aug, adj=adj_aug)
            js_v = rowwise_js_divergence(base_probs, out_v["role_probs"])
            js_list.append(js_v.detach().cpu().numpy())

    js_mean = np.mean(np.vstack(js_list), axis=0)
    js_norm = np.clip(js_mean / math.log(2.0), 0.0, 1.0)
    role_stability = 1.0 - js_norm
    return np.clip(role_stability, 0.0, 1.0)

def map_role_names(role_weight_matrix: np.ndarray) -> List[str]:
    component_names = ["structural", "semantic", "perturbation", "contextual", "non_redundancy"]
    dominant_to_name = {
        "structural": "bridge",
        "semantic": "driver",
        "perturbation": "stabilizer",
        "contextual": "contextualizer",
        "non_redundancy": "unique_survivor",
    }

    names = []
    for idx, row in enumerate(role_weight_matrix):
        dominant_component = component_names[int(np.argmax(row))]
        names.append(f"{dominant_to_name[dominant_component]}_role_{idx + 1}")
    return names

def build_feature_report(
    component_df: pd.DataFrame,
    gnn_out,
    model: nn.Module,
    role_stability: np.ndarray,
) -> pd.DataFrame:
    comp_cols = ["structural", "semantic", "perturbation", "contextual", "non_redundancy"]

    role_probs = gnn_out["role_probs"].detach().cpu().numpy()
    role_persistence = gnn_out["role_persistence"].detach().cpu().numpy()
    omega = gnn_out["omega"].detach().cpu().numpy()
    role_weight_matrix = gnn_out["role_weight_matrix"].detach().cpu().numpy()
    component_pred = gnn_out["component_pred"].detach().cpu().numpy()

    role_names = map_role_names(role_weight_matrix)
    top_role_idx = role_probs.argmax(axis=1)
    top_role_name = [role_names[i] for i in top_role_idx]

    core_evidence = component_df[comp_cols].values.astype(float)
    omega_dot_c = (omega * core_evidence).sum(axis=1)

    gamma = float(model.gamma.detach().cpu().item())
    eta = float(model.eta.detach().cpu().item())
    mu = float(model.mu.detach().cpu().item())
    bias = float(model.bias.detach().cpu().item())

    final_logit = (
        gamma * omega_dot_c
        + ROLE_PERSISTENCE_BETA * role_persistence
        + ROLE_STABILITY_WEIGHT * role_stability
        + eta * component_df["topology_persistence"].values
        + mu * component_df["substitute_margin"].values
        + bias
    )
    csi_score = sigmoid_np(final_logit)

    report = component_df.copy()
    report["role_persistence"] = role_persistence
    report["role_stability"] = role_stability
    report["omega_dot_c"] = omega_dot_c
    report["cts_score"] = csi_score
    report["top_role"] = top_role_name

    for i, col in enumerate(comp_cols):
        report[f"omega_{col}"] = omega[:, i]
        report[f"pred_{col}"] = component_pred[:, i]

    for i in range(role_probs.shape[1]):
        report[f"role_prob_{i+1}"] = role_probs[:, i]

    report = report.sort_values("cts_score", ascending=False).reset_index(drop=True)
    return report

def select_features(
    feature_report: pd.DataFrame,
    corr_matrix: pd.DataFrame,
) -> List[str]:
    ranked = feature_report.sort_values("cts_score", ascending=False).copy()

    candidate = ranked.loc[ranked["cts_score"] >= SURVIVAL_THRESHOLD, "feature"].tolist()

    min_keep = min(MIN_FEATURES_TO_KEEP, len(ranked))
    max_keep = min(MAX_FEATURES_TO_KEEP, len(ranked))

    if len(candidate) < min_keep:
        candidate = ranked["feature"].tolist()[:max(min_keep, len(candidate))]
    if len(candidate) > max_keep:
        candidate = candidate[:max_keep]

    selected = []

    role_seeded = 0
    seen_roles = set()
    for _, row in ranked.iterrows():
        feat = row["feature"]
        role = row["top_role"]
        if feat not in candidate:
            continue
        if role not in seen_roles:
            selected.append(feat)
            seen_roles.add(role)
            role_seeded += 1
        if role_seeded >= MIN_ROLE_DIVERSE_SEEDS * min(len(seen_roles) + 1, N_ROLES):
            break

    for feat in candidate:
        if feat in selected:
            continue
        if not selected:
            selected.append(feat)
            continue
        max_corr = float(corr_matrix.loc[feat, selected].max()) if selected else 0.0
        if max_corr < REDUNDANCY_THRESHOLD:
            selected.append(feat)
        if len(selected) >= max_keep:
            break

    if len(selected) < min_keep:
        for feat in ranked["feature"].tolist():
            if feat not in selected:
                selected.append(feat)
            if len(selected) >= min_keep:
                break

    return selected[:max_keep]

def save_json(data: Dict, path: str) -> None:
    with open(path, "w") as f:
        json.dump(data, f, indent=2)

def run_grafts_feature_selection():
    seed_everything(RANDOM_STATE)
    ensure_dir(OUTPUT_DIR)

    print_rule("Loading train/test data")
    train_df = read_table(TRAIN_PATH)
    test_df = read_table(TEST_PATH)

    if TARGET_COLUMN not in train_df.columns:
        raise ValueError(f"TARGET_COLUMN '{TARGET_COLUMN}' not found in train file.")

    original_feature_cols = [c for c in train_df.columns if c not in set(ID_COLUMNS + [TARGET_COLUMN])]

    if test_df is not None:
        missing_in_test = [c for c in original_feature_cols if c not in test_df.columns]
        if missing_in_test:
            raise ValueError(f"Test file is missing these train feature columns: {missing_in_test}")

    usable_features = []
    dropped_constant = []
    for col in original_feature_cols:
        nunq = train_df[col].nunique(dropna=True)
        if nunq <= 1:
            dropped_constant.append(col)
        else:
            usable_features.append(col)

    feature_cols = usable_features
    task_type = TASK_TYPE if TASK_TYPE != "auto" else infer_task_type(train_df[TARGET_COLUMN])

    y_train_full_raw = train_df[TARGET_COLUMN].copy()
    if task_type == "classification":
        train_labels = y_train_full_raw.astype("string")
        unique_train_labels = sorted(train_labels.dropna().unique().tolist())
        label_mapping = {label: idx for idx, label in enumerate(unique_train_labels)}
        y_train_full = train_labels.map(label_mapping).astype(int).values
    else:
        label_mapping = None
        y_train_full = pd.to_numeric(y_train_full_raw, errors="coerce").values.astype(float)

    X_full = train_df[feature_cols].copy()

    if task_type == "classification":
        X_train_inner, X_valid_inner, y_train_inner, y_valid_inner = train_test_split(
            X_full,
            y_train_full,
            test_size=VALID_SIZE,
            random_state=RANDOM_STATE,
            stratify=y_train_full,
        )
    else:
        X_train_inner, X_valid_inner, y_train_inner, y_valid_inner = train_test_split(
            X_full,
            y_train_full,
            test_size=VALID_SIZE,
            random_state=RANDOM_STATE,
        )

    print(f"Task type               : {task_type}")
    print(f"Train rows              : {len(train_df)}")
    print(f"Test rows               : {0 if test_df is None else len(test_df)}")
    print(f"Original feature count  : {len(original_feature_cols)}")
    print(f"Usable feature count    : {len(feature_cols)}")
    print(f"Dropped constant feats  : {len(dropped_constant)}")
    print(f"Device                  : {DEVICE}")

    print_rule("Leakage-safe inner-train encoding")
    encoder_inner = LeakageSafeTabularEncoder()
    X_train_enc = encoder_inner.fit_transform(X_train_inner)
    X_valid_enc = encoder_inner.transform(X_valid_inner)

    print_rule("Dual graph construction")
    corr_matrix, miss_corr_matrix, A_stat, A_stat_norm = build_statistical_graph(
        X_train_raw=X_train_inner,
        X_train_enc=X_train_enc,
        top_k=TOP_K_NEIGHBORS,
        alpha=STAT_GRAPH_ALPHA,
    )

    print(f"Stat graph nodes        : {A_stat.shape[0]}")
    print(f"Stat graph edges        : {int((A_stat > 0).sum() // 2)}")

    print_rule("Multi-evidence component computation")
    component_df_seed, response_profile_df, base_model = compute_observed_components_and_profiles(
        X_train_raw=X_train_inner,
        X_train_enc=X_train_enc,
        y_train=y_train_inner,
        X_valid_enc=X_valid_enc,
        y_valid=y_valid_inner,
        corr_matrix=corr_matrix,
        A_fused=A_stat,
        task_type=task_type,
    )

    A_resp, A_resp_norm = build_response_graph(
        response_profile_df=response_profile_df,
        top_k=TOP_K_NEIGHBORS,
    )
    A_fused, A_fused_norm = fuse_graphs(A_stat=A_stat, A_resp=A_resp, alpha=FUSED_GRAPH_ALPHA)

    print(f"Resp graph edges        : {int((A_resp > 0).sum() // 2)}")
    print(f"Fused graph edges       : {int((A_fused > 0).sum() // 2)}")

    print_rule("Recompute evidence on fused graph")
    component_df, response_profile_df, _ = compute_observed_components_and_profiles(
        X_train_raw=X_train_inner,
        X_train_enc=X_train_enc,
        y_train=y_train_inner,
        X_valid_enc=X_valid_enc,
        y_valid=y_valid_inner,
        corr_matrix=corr_matrix,
        A_fused=A_fused,
        task_type=task_type,
    )

    print_rule("Build node feature matrix")
    node_df = build_node_feature_matrix(
        X_train_raw=X_train_inner,
        X_train_enc=X_train_enc,
        y_train=y_train_inner,
        corr_matrix=corr_matrix,
        A_stat=A_stat,
        A_resp=A_resp,
        A_fused=A_fused,
        component_df=component_df,
        task_type=task_type,
    )

    print_rule("Train GRAFTS GNN")
    gnn_model, gnn_out = train_grafts_gnn(
        node_features=node_df,
        A_fused_norm=A_fused_norm,
        component_df=component_df,
    )

    print_rule("Compute role stability")
    role_stability = compute_role_stability(
        model=gnn_model,
        node_features=node_df,
        A_fused_norm=A_fused_norm,
        n_views=N_ROLE_STABILITY_VIEWS,
    )

    print_rule("Build feature report + select features")
    feature_report = build_feature_report(
        component_df=component_df,
        gnn_out=gnn_out,
        model=gnn_model,
        role_stability=role_stability,
    )
    selected_features = select_features(feature_report, corr_matrix)

    print(f"Selected feature count  : {len(selected_features)}")

    print_rule("Compare all-features vs selected-features on inner validation")
    eval_all = evaluate_subset(
        task_type=task_type,
        X_train=X_train_enc,
        y_train=y_train_inner,
        X_valid=X_valid_enc,
        y_valid=y_valid_inner,
        random_state=RANDOM_STATE,
        n_trees=RF_TREES,
    )
    eval_sel = evaluate_subset(
        task_type=task_type,
        X_train=X_train_enc[selected_features],
        y_train=y_train_inner,
        X_valid=X_valid_enc[selected_features],
        y_valid=y_valid_inner,
        random_state=RANDOM_STATE,
        n_trees=RF_TREES,
    )

    compare_df = pd.DataFrame([
        {"subset": "all_features", **eval_all},
        {"subset": "selected_features", **eval_sel},
    ])

    print_rule("Selected feature report")
    display(feature_report.head(20))
    display(
        feature_report[feature_report["feature"].isin(selected_features)]
        .sort_values("cts_score", ascending=False)
        .reset_index(drop=True)
    )
    display(compare_df)

    print_rule("Saving outputs")
    feature_report_path = os.path.join(OUTPUT_DIR, "grafts_feature_report.csv")
    selected_features_path = os.path.join(OUTPUT_DIR, "selected_features.csv")
    train_selected_path = os.path.join(OUTPUT_DIR, "train_selected.csv")
    test_selected_path = os.path.join(OUTPUT_DIR, "test_selected.csv")
    encoder_state_path = os.path.join(OUTPUT_DIR, "inner_encoder_state.joblib")
    config_path = os.path.join(OUTPUT_DIR, "fs_config_used.json")
    node_features_path = os.path.join(OUTPUT_DIR, "node_features.csv")
    compare_path = os.path.join(OUTPUT_DIR, "subset_comparison.csv")
    response_profile_path = os.path.join(OUTPUT_DIR, "response_profiles.csv")
    graph_stat_path = os.path.join(OUTPUT_DIR, "graph_stat.npy")
    graph_resp_path = os.path.join(OUTPUT_DIR, "graph_response.npy")
    graph_fused_path = os.path.join(OUTPUT_DIR, "graph_fused.npy")

    feature_report.to_csv(feature_report_path, index=False)
    pd.DataFrame({"selected_feature": selected_features}).to_csv(selected_features_path, index=False)
    node_df.to_csv(node_features_path)
    response_profile_df.to_csv(response_profile_path)
    compare_df.to_csv(compare_path, index=False)

    np.save(graph_stat_path, A_stat)
    np.save(graph_resp_path, A_resp)
    np.save(graph_fused_path, A_fused)

    train_keep_cols = unique_preserve_order(
        ID_COLUMNS + selected_features + ([TARGET_COLUMN] if TARGET_COLUMN in train_df.columns else [])
    )
    train_selected_df = train_df[train_keep_cols].copy()

    if test_df is not None:
        test_keep_cols = unique_preserve_order(
            ID_COLUMNS + selected_features + ([TARGET_COLUMN] if TARGET_COLUMN in test_df.columns else [])
        )
        test_keep_cols = [c for c in test_keep_cols if c in test_df.columns]
        test_selected_df = test_df[test_keep_cols].copy()
        test_selected_df.to_csv(test_selected_path, index=False)
    else:
        test_selected_df = None

    train_selected_df.to_csv(train_selected_path, index=False)
    train_selected_df.to_csv("/content/train_selected.csv", index=False)
    if test_selected_df is not None:
        test_selected_df.to_csv("/content/test_selected.csv", index=False)
    joblib.dump(encoder_inner.get_state(), encoder_state_path)

    config_to_save = {
        "METHOD_NAME": "GRAFTS",
        "TRAIN_PATH": TRAIN_PATH,
        "TEST_PATH": TEST_PATH,
        "TARGET_COLUMN": TARGET_COLUMN,
        "ID_COLUMNS": ID_COLUMNS,
        "OUTPUT_DIR": OUTPUT_DIR,
        "TASK_TYPE": task_type,
        "RANDOM_STATE": RANDOM_STATE,
        "VALID_SIZE": VALID_SIZE,
        "TOP_K_NEIGHBORS": TOP_K_NEIGHBORS,
        "N_ROLES": N_ROLES,
        "HIDDEN_DIM": HIDDEN_DIM,
        "GNN_EPOCHS": GNN_EPOCHS,
        "GNN_LR": GNN_LR,
        "ROLE_PERSISTENCE_BETA": ROLE_PERSISTENCE_BETA,
        "ROLE_STABILITY_WEIGHT": ROLE_STABILITY_WEIGHT,
        "RF_TREES": RF_TREES,
        "N_PERTURB_TRIALS": N_PERTURB_TRIALS,
        "N_CONTEXT_CLUSTERS": N_CONTEXT_CLUSTERS,
        "MIN_VALID_CLUSTER_SIZE": MIN_VALID_CLUSTER_SIZE,
        "STAT_GRAPH_ALPHA": STAT_GRAPH_ALPHA,
        "FUSED_GRAPH_ALPHA": FUSED_GRAPH_ALPHA,
        "N_TOPOLOGY_VIEWS": N_TOPOLOGY_VIEWS,
        "GRAPH_EDGE_DROP_RATE": GRAPH_EDGE_DROP_RATE,
        "N_ROLE_STABILITY_VIEWS": N_ROLE_STABILITY_VIEWS,
        "NODE_NOISE_STD": NODE_NOISE_STD,
        "SURVIVAL_THRESHOLD": SURVIVAL_THRESHOLD,
        "REDUNDANCY_THRESHOLD": REDUNDANCY_THRESHOLD,
        "MIN_FEATURES_TO_KEEP": MIN_FEATURES_TO_KEEP,
        "MAX_FEATURES_TO_KEEP": MAX_FEATURES_TO_KEEP,
        "MIN_ROLE_DIVERSE_SEEDS": MIN_ROLE_DIVERSE_SEEDS,
        "RANKING_MARGIN": RANKING_MARGIN,
        "MAX_RANK_PAIRS": MAX_RANK_PAIRS,
        "DROPPED_CONSTANT_FEATURES": dropped_constant,
        "SELECTED_FEATURES": selected_features,
        "LABEL_MAPPING": label_mapping,
        "INNER_VALIDATION_SCORE_ALL": eval_all,
        "INNER_VALIDATION_SCORE_SELECTED": eval_sel,
    }
    save_json(config_to_save, config_path)

    print(f"Saved: {feature_report_path}")
    print(f"Saved: {selected_features_path}")
    print(f"Saved: {train_selected_path}")
    if test_df is not None:
        print(f"Saved: {test_selected_path}")
    print(f"Saved: {encoder_state_path}")
    print(f"Saved: {node_features_path}")
    print(f"Saved: {response_profile_path}")
    print(f"Saved: {compare_path}")
    print(f"Saved: {graph_stat_path}")
    print(f"Saved: {graph_resp_path}")
    print(f"Saved: {graph_fused_path}")
    print(f"Saved: {config_path}")

    print_rule("Final selected feature names")
    for feat in selected_features:
        print(f"- {feat}")

    return {
        "feature_report": feature_report,
        "selected_features": selected_features,
        "train_selected_df": train_selected_df,
        "test_selected_df": test_selected_df,
        "subset_compare_df": compare_df,
        "node_features_df": node_df,
        "response_profile_df": response_profile_df,
    }

fs_results = run_grafts_feature_selection()
```

```python
import os
import json
import time
import copy
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from collections import Counter

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler, label_binarize
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    log_loss,
    matthews_corrcoef,
    confusion_matrix,
    balanced_accuracy_score,
    brier_score_loss,
)

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

warnings.filterwarnings("ignore")

TRAIN_PATH = "/content/train_selected.csv"
TEST_PATH  = "/content/test_selected.csv"
TARGET_COLUMN = "diagnosis"
ID_COLUMNS = []
DROP_COLUMNS = ["Patient_ID"]
OUTPUT_DIR = "/content/grafts_tabfgt_outputs"
SAVE_OUTPUTS = True
SHOW_PLOTS = False
SEED = 42
OUTER_FOLDS = 5
INNER_FOLDS = 2
N_TRIALS = 3
MAX_EPOCHS_INNER = 6
MAX_EPOCHS_OUTER = 10
MAX_EPOCHS_FINAL = 12

PATIENCE_INNER = 2
PATIENCE_OUTER = 3
PATIENCE_FINAL = 3

OUTER_DEV_VALID_SIZE = 0.12
FINAL_VALID_SIZE = 0.10

USE_AMP = True
NUM_WORKERS = 0
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
THRESHOLD_GRID = np.linspace(0.16, 0.60, 177)
ACC_DROP_TOL = 0.0015
MIN_DYNAMIC_PRECISION = 0.970
PRECISION_RELAX = 0.012
USE_MILD_POS_WEIGHT = True
POS_WEIGHT_POWER = 0.35
POS_WEIGHT_MAX = 1.25
HP_ACC_TOL = 0.0015
FINAL_ENSEMBLE_SEEDS = [42]

SEARCH_CANDIDATES = [
    {"embed_dim": 128, "num_heads": 4, "depth": 2, "ff_mult": 2.0, "dropout": 0.08, "token_dropout": 0.01, "lr": 1.0e-3, "weight_decay": 1e-5, "batch_size": 64},
    {"embed_dim": 160, "num_heads": 4, "depth": 2, "ff_mult": 2.0, "dropout": 0.10, "token_dropout": 0.02, "lr": 8.0e-4, "weight_decay": 1e-5, "batch_size": 64},
    {"embed_dim": 192, "num_heads": 4, "depth": 3, "ff_mult": 2.0, "dropout": 0.10, "token_dropout": 0.02, "lr": 8.0e-4, "weight_decay": 1e-5, "batch_size": 64},
]

IS_BINARY = True
N_CLASSES = 2
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

    if torch.backends.cudnn.is_available():
        torch.backends.cudnn.deterministic = False
        torch.backends.cudnn.benchmark = True

    if torch.cuda.is_available():
        torch.backends.cuda.matmul.allow_tf32 = True
        torch.backends.cudnn.allow_tf32 = True

    try:
        torch.set_float32_matmul_precision("high")
    except Exception:
        pass

def now():
    return time.strftime("%H:%M:%S")

def ensure_dir(path):
    os.makedirs(path, exist_ok=True)

def safe_read_csv(path):
    if path is None or str(path).strip() == "":
        return None
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing file: {path}")
    return pd.read_csv(path)

def to_numeric_df(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c], errors="coerce")
    return out

def clip_probs(prob, eps=1e-7):
    prob = np.asarray(prob, dtype=np.float64)
    prob = np.clip(prob, eps, 1 - eps)
    if prob.ndim == 2:
        row_sum = prob.sum(axis=1, keepdims=True)
        row_sum = np.clip(row_sum, eps, None)
        prob = prob / row_sum
    return prob

def safe_auc(y_true, prob):
    prob = np.asarray(prob)
    try:
        if prob.ndim == 1:
            if len(np.unique(y_true)) < 2:
                return np.nan
            return float(roc_auc_score(y_true, prob))
        else:
            if len(np.unique(y_true)) < 2:
                return np.nan
            return float(
                roc_auc_score(
                    y_true,
                    prob,
                    multi_class="ovr",
                    average="macro",
                    labels=np.arange(prob.shape[1]),
                )
            )
    except Exception:
        return np.nan

def safe_pr_auc(y_true, prob):
    prob = np.asarray(prob)
    try:
        if prob.ndim == 1:
            if len(np.unique(y_true)) < 2:
                return np.nan
            return float(average_precision_score(y_true, prob))
        else:
            if len(np.unique(y_true)) < 2:
                return np.nan
            y_bin = label_binarize(y_true, classes=np.arange(prob.shape[1]))
            return float(average_precision_score(y_bin, prob, average="macro"))
    except Exception:
        return np.nan

def multiclass_specificity(y_true, y_pred, n_classes):
    cm = confusion_matrix(y_true, y_pred, labels=np.arange(n_classes))
    total = cm.sum()
    specs = []

    for i in range(n_classes):
        tp = cm[i, i]
        fp = cm[:, i].sum() - tp
        fn = cm[i, :].sum() - tp
        tn = total - tp - fp - fn
        spec_i = tn / max(tn + fp, 1)
        specs.append(spec_i)

    return float(np.mean(specs))

def multiclass_brier_score(y_true, prob, n_classes):
    y_bin = label_binarize(y_true, classes=np.arange(n_classes))
    prob = clip_probs(prob)
    return float(np.mean(np.sum((y_bin - prob) ** 2, axis=1)))

def metric_score_binary(y_true, y_pred, metric_name="accuracy"):
    if metric_name == "accuracy":
        return accuracy_score(y_true, y_pred)
    if metric_name == "f1":
        return f1_score(y_true, y_pred, zero_division=0)
    if metric_name == "mcc":
        return matthews_corrcoef(y_true, y_pred)
    if metric_name == "balanced_accuracy":
        return balanced_accuracy_score(y_true, y_pred)
    raise ValueError(f"Unsupported metric: {metric_name}")

def _threshold_table(y_true, prob, grid):
    rows = []
    prob = clip_probs(prob)

    for t in grid:
        pred = (prob >= t).astype(int)
        acc = accuracy_score(y_true, pred)
        prec = precision_score(y_true, pred, zero_division=0)
        rec = recall_score(y_true, pred, zero_division=0)
        f1v = f1_score(y_true, pred, zero_division=0)
        mcc = matthews_corrcoef(y_true, pred)

        rows.append({
            "threshold": float(t),
            "accuracy": float(acc),
            "precision": float(prec),
            "recall": float(rec),
            "f1": float(f1v),
            "mcc": float(mcc),
        })

    return pd.DataFrame(rows)

def _pick_best_threshold_from_df(df_thr: pd.DataFrame) -> float:
    if df_thr.empty:
        return 0.50

    df_thr = df_thr.sort_values(
        by=["f1", "recall", "accuracy", "mcc", "precision", "threshold"],
        ascending=[False, False, False, False, False, True]
    ).reset_index(drop=True)

    return float(df_thr.loc[0, "threshold"])

def tune_threshold_constrained(
    y_true,
    prob,
    grid=None,
    acc_drop_tol=0.0015,
    min_dynamic_precision=0.970,
    precision_relax=0.012,
):
    if grid is None:
        grid = np.linspace(0.16, 0.60, 177)

    prob = clip_probs(prob)
    thr_df = _threshold_table(y_true, prob, grid)

    best_acc = float(thr_df["accuracy"].max())
    acc_floor = best_acc - float(acc_drop_tol)

    safe_df = thr_df.loc[thr_df["accuracy"] >= acc_floor].copy()
    if safe_df.empty:
        safe_df = thr_df.copy()

    best_acc_df = thr_df.loc[thr_df["accuracy"] >= best_acc - 1e-12].copy()
    best_acc_df = best_acc_df.sort_values(
        by=["f1", "recall", "mcc", "precision", "threshold"],
        ascending=[False, False, False, False, True]
    ).reset_index(drop=True)

    ref_precision = float(best_acc_df.loc[0, "precision"])
    precision_floor = max(float(min_dynamic_precision), ref_precision - float(precision_relax))

    guarded_df = safe_df.loc[safe_df["precision"] >= precision_floor].copy()
    if guarded_df.empty:
        guarded_df = safe_df.copy()

    return _pick_best_threshold_from_df(guarded_df)

def choose_best_hp_info(info_list, acc_tol=0.0015):
    rows = []
    for i, info in enumerate(info_list):
        m = info["metrics"]
        rows.append({
            "idx": i,
            "accuracy": float(m["accuracy"]),
            "f1": float(m["f1"]),
            "recall": float(m["recall"]),
            "precision": float(m["precision"]),
            "mcc": float(m["mcc"]),
            "logloss": float(m["logloss"]),
        })

    df = pd.DataFrame(rows)
    best_acc = float(df["accuracy"].max())

    keep = df.loc[df["accuracy"] >= best_acc - float(acc_tol)].copy()
    keep = keep.sort_values(
        by=["f1", "recall", "mcc", "precision", "accuracy", "logloss"],
        ascending=[False, False, False, False, False, True]
    ).reset_index(drop=True)

    return info_list[int(keep.loc[0, "idx"])]

def compute_metrics(y_true, prob, threshold=0.5):
    prob = np.asarray(prob)

    if prob.ndim == 1:
        prob = clip_probs(prob)
        pred = (prob >= threshold).astype(int)

        cm = confusion_matrix(y_true, pred, labels=[0, 1])
        tn, fp, fn, tp = cm.ravel()
        specificity = tn / max(tn + fp, 1)

        return {
            "threshold": float(threshold),
            "accuracy": float(accuracy_score(y_true, pred)),
            "precision": float(precision_score(y_true, pred, zero_division=0)),
            "recall": float(recall_score(y_true, pred, zero_division=0)),
            "specificity": float(specificity),
            "f1": float(f1_score(y_true, pred, zero_division=0)),
            "mcc": float(matthews_corrcoef(y_true, pred)),
            "balanced_accuracy": float(balanced_accuracy_score(y_true, pred)),
            "roc_auc": safe_auc(y_true, prob),
            "pr_auc": safe_pr_auc(y_true, prob),
            "logloss": float(log_loss(y_true, prob, labels=[0, 1])),
            "brier": float(brier_score_loss(y_true, prob)),
            "tn": int(tn),
            "fp": int(fp),
            "fn": int(fn),
            "tp": int(tp),
        }

    prob = clip_probs(prob)
    pred = np.argmax(prob, axis=1)
    n_classes = prob.shape[1]
    cm = confusion_matrix(y_true, pred, labels=np.arange(n_classes))

    return {
        "threshold": None,
        "accuracy": float(accuracy_score(y_true, pred)),
        "precision": float(precision_score(y_true, pred, average="macro", zero_division=0)),
        "recall": float(recall_score(y_true, pred, average="macro", zero_division=0)),
        "specificity": float(multiclass_specificity(y_true, pred, n_classes)),
        "f1": float(f1_score(y_true, pred, average="macro", zero_division=0)),
        "mcc": float(matthews_corrcoef(y_true, pred)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true, pred)),
        "roc_auc": safe_auc(y_true, prob),
        "pr_auc": safe_pr_auc(y_true, prob),
        "logloss": float(log_loss(y_true, prob, labels=np.arange(n_classes))),
        "brier": float(multiclass_brier_score(y_true, prob, n_classes)),
        "tn": None,
        "fp": None,
        "fn": None,
        "tp": None,
    }

def plot_conf_matrix(cm, title="Confusion Matrix", class_names=None):
    n = cm.shape[0]
    if class_names is None:
        class_names = [str(i) for i in range(n)]

    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(cm, cmap="Blues")
    plt.colorbar(im, ax=ax)
    ax.set_xticks(np.arange(n))
    ax.set_yticks(np.arange(n))
    ax.set_xticklabels(class_names, rotation=45, ha="right")
    ax.set_yticklabels(class_names)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.set_title(title)

    thresh = cm.max() / 2 if cm.max() > 0 else 0.5
    for i in range(n):
        for j in range(n):
            ax.text(
                j, i, str(cm[i, j]),
                ha="center", va="center",
                color="white" if cm[i, j] > thresh else "black"
            )
    plt.tight_layout()
    plt.show()

def hp_to_str(hp):
    return (
        f"embed={hp['embed_dim']} | heads={hp['num_heads']} | depth={hp['depth']} | "
        f"ff_mult={hp['ff_mult']} | drop={hp['dropout']} | tok_drop={hp['token_dropout']} | "
        f"lr={hp['lr']} | wd={hp['weight_decay']} | bs={hp['batch_size']}"
    )

class FoldPreprocessor:
    def __init__(self):
        self.imputer = SimpleImputer(strategy="median")
        self.scaler = StandardScaler()

    def fit(self, X_df):
        X_num = to_numeric_df(X_df)
        X_imp = self.imputer.fit_transform(X_num)
        self.scaler.fit(X_imp)
        return self

    def transform(self, X_df):
        X_num = to_numeric_df(X_df)
        X_imp = self.imputer.transform(X_num)
        X_scaled = self.scaler.transform(X_imp)
        X_scaled = np.clip(X_scaled, -8.0, 8.0).astype(np.float32)
        return X_scaled

    def fit_transform(self, X_df):
        self.fit(X_df)
        return self.transform(X_df)

class NumericFeatureTokenizer(nn.Module):
    def __init__(self, n_features, embed_dim, token_dropout=0.01):
        super().__init__()
        self.n_features = n_features
        self.embed_dim = embed_dim
        self.token_dropout = float(token_dropout)

        self.weight = nn.Parameter(torch.randn(n_features, embed_dim) * 0.02)
        self.bias = nn.Parameter(torch.zeros(n_features, embed_dim))
        self.feature_embed = nn.Parameter(torch.randn(n_features, embed_dim) * 0.02)
        self.norm = nn.LayerNorm(embed_dim)

    def forward(self, x):
        tokens = x.unsqueeze(-1) * self.weight.unsqueeze(0) + self.bias.unsqueeze(0) + self.feature_embed.unsqueeze(0)

        if self.training and self.token_dropout > 0:
            keep = (torch.rand((x.size(0), x.size(1), 1), device=x.device) > self.token_dropout).float()
            tokens = tokens * keep

        tokens = self.norm(tokens)
        return tokens

class TransformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, ff_mult=2.0, dropout=0.10):
        super().__init__()
        ff_dim = int(embed_dim * ff_mult)

        self.norm1 = nn.LayerNorm(embed_dim)
        self.attn = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )
        self.drop1 = nn.Dropout(dropout)

        self.norm2 = nn.LayerNorm(embed_dim)
        self.ff1 = nn.Linear(embed_dim, ff_dim * 2)
        self.ff2 = nn.Linear(ff_dim, embed_dim)
        self.drop2 = nn.Dropout(dropout)

        self.gate = nn.Sequential(
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, embed_dim),
            nn.Sigmoid()
        )

    def forward(self, x):
        y = self.norm1(x)
        attn_out, _ = self.attn(y, y, y, need_weights=False)
        x = x + self.drop1(attn_out)

        y = self.norm2(x)
        a, b = self.ff1(y).chunk(2, dim=-1)
        y = a * F.gelu(b)
        y = self.ff2(self.drop2(y))

        gate = self.gate(x)
        x = x + gate * y
        return x

class TabFGT(nn.Module):
    def __init__(
        self,
        input_dim,
        num_classes=1,
        embed_dim=160,
        num_heads=4,
        depth=2,
        ff_mult=2.0,
        dropout=0.10,
        token_dropout=0.01,
    ):
        super().__init__()

        self.num_classes = num_classes

        self.tokenizer = NumericFeatureTokenizer(
            n_features=input_dim,
            embed_dim=embed_dim,
            token_dropout=token_dropout
        )

        self.cls_token = nn.Parameter(torch.zeros(1, 1, embed_dim))
        self.cls_bias = nn.Parameter(torch.zeros(1, 1, embed_dim))

        self.blocks = nn.ModuleList([
            TransformerBlock(
                embed_dim=embed_dim,
                num_heads=num_heads,
                ff_mult=ff_mult,
                dropout=dropout
            )
            for _ in range(depth)
        ])

        self.final_norm = nn.LayerNorm(embed_dim)

        head_dim = embed_dim * 2
        hidden2 = max(embed_dim, 128)

        self.head = nn.Sequential(
            nn.LayerNorm(head_dim),
            nn.Linear(head_dim, hidden2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden2, hidden2 // 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden2 // 2, num_classes),
        )

    def forward(self, x):
        tok = self.tokenizer(x)
        cls = self.cls_token.expand(x.size(0), -1, -1) + self.cls_bias
        tok = torch.cat([cls, tok], dim=1)

        for block in self.blocks:
            tok = block(tok)

        tok = self.final_norm(tok)

        cls_tok = tok[:, 0]
        feat_tok = tok[:, 1:]
        mean_tok = feat_tok.mean(dim=1)

        rep = torch.cat([cls_tok, mean_tok], dim=1)
        out = self.head(rep)

        if self.num_classes == 1:
            out = out.squeeze(1)

        return out

def make_loader(X, y=None, batch_size=1024, shuffle=False):
    X_t = torch.tensor(X, dtype=torch.float32)
    if y is None:
        ds = TensorDataset(X_t)
    else:
        if IS_BINARY:
            y_t = torch.tensor(y, dtype=torch.float32)
        else:
            y_t = torch.tensor(y, dtype=torch.long)
        ds = TensorDataset(X_t, y_t)

    return DataLoader(
        ds,
        batch_size=batch_size,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=(DEVICE.type == "cuda"),
        drop_last=False,
    )

@torch.no_grad()
def predict_proba(model, X, batch_size=1024):
    model.eval()
    loader = make_loader(X, y=None, batch_size=batch_size, shuffle=False)
    probs = []

    for batch in loader:
        xb = batch[0].to(DEVICE, non_blocking=True)
        logits = model(xb)

        if IS_BINARY:
            p = torch.sigmoid(logits).detach().cpu().numpy()
        else:
            p = torch.softmax(logits, dim=1).detach().cpu().numpy()

        probs.append(p)

    if IS_BINARY:
        return np.concatenate(probs).astype(np.float64)
    return np.vstack(probs).astype(np.float64)

def train_one_model(
    X_train, y_train,
    X_valid, y_valid,
    hp,
    max_epochs=12,
    patience=3,
    seed=42
):
    set_seed(seed)

    model = TabFGT(
        input_dim=X_train.shape[1],
        num_classes=(1 if IS_BINARY else N_CLASSES),
        embed_dim=hp["embed_dim"],
        num_heads=hp["num_heads"],
        depth=hp["depth"],
        ff_mult=hp["ff_mult"],
        dropout=hp["dropout"],
        token_dropout=hp["token_dropout"],
    ).to(DEVICE)

    train_loader = make_loader(X_train, y_train, batch_size=hp["batch_size"], shuffle=True)
    valid_loader = make_loader(X_valid, y_valid, batch_size=max(1024, hp["batch_size"]), shuffle=False)

    if IS_BINARY:
        pos_count = max(float(np.sum(np.asarray(y_train) == 1)), 1.0)
        neg_count = max(float(np.sum(np.asarray(y_train) == 0)), 1.0)

        if USE_MILD_POS_WEIGHT:
            pos_weight_value = float(np.clip((neg_count / pos_count) ** POS_WEIGHT_POWER, 1.0, POS_WEIGHT_MAX))
        else:
            pos_weight_value = 1.0

        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight_value], dtype=torch.float32, device=DEVICE)
        )

    else:
        counts = np.bincount(np.asarray(y_train).astype(int), minlength=N_CLASSES).astype(np.float64)
        max_count = max(float(counts.max()), 1.0)

        if USE_MILD_POS_WEIGHT:
            class_weights = (max_count / np.maximum(counts, 1.0)) ** POS_WEIGHT_POWER
            class_weights = np.clip(class_weights, 1.0, POS_WEIGHT_MAX)
        else:
            class_weights = np.ones(N_CLASSES, dtype=np.float64)

        pos_weight_value = float(np.mean(class_weights))
        criterion = nn.CrossEntropyLoss(
            weight=torch.tensor(class_weights, dtype=torch.float32, device=DEVICE)
        )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=hp["lr"],
        weight_decay=hp["weight_decay"]
    )

    steps_per_epoch = max(1, len(train_loader))
    total_steps = max(1, max_epochs * steps_per_epoch)

    scheduler = torch.optim.lr_scheduler.OneCycleLR(
        optimizer,
        max_lr=hp["lr"],
        total_steps=total_steps,
        pct_start=0.20,
        div_factor=8.0,
        final_div_factor=80.0,
    )

    use_amp_now = bool(USE_AMP and DEVICE.type == "cuda")
    scaler = torch.cuda.amp.GradScaler(enabled=use_amp_now)

    best_state = copy.deepcopy(model.state_dict())
    best_val_loss = np.inf
    best_epoch = 0
    bad_epochs = 0
    min_delta = 5e-4

    for epoch in range(1, max_epochs + 1):
        model.train()

        for xb, yb in train_loader:
            xb = xb.to(DEVICE, non_blocking=True)
            yb = yb.to(DEVICE, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            with torch.cuda.amp.autocast(enabled=use_amp_now):
                logits = model(xb)
                loss = criterion(logits, yb)

            scaler.scale(loss).backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()
            scheduler.step()

        model.eval()
        val_probs = []

        for xb, _ in valid_loader:
            xb = xb.to(DEVICE, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=use_amp_now):
                logits = model(xb)
                if IS_BINARY:
                    p = torch.sigmoid(logits)
                else:
                    p = torch.softmax(logits, dim=1)
            val_probs.append(p.detach().cpu().numpy())

        if IS_BINARY:
            val_probs = np.concatenate(val_probs).astype(np.float64)
            val_loss = log_loss(y_valid, clip_probs(val_probs), labels=[0, 1])
        else:
            val_probs = np.vstack(val_probs).astype(np.float64)
            val_loss = log_loss(y_valid, clip_probs(val_probs), labels=np.arange(N_CLASSES))

        if val_loss < best_val_loss - min_delta:
            best_val_loss = float(val_loss)
            best_epoch = int(epoch)
            best_state = copy.deepcopy(model.state_dict())
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    model.load_state_dict(best_state)
    return model, best_epoch, best_val_loss, pos_weight_value

def inner_cv_score(X_df, y, hp, seed=42):
    skf = StratifiedKFold(n_splits=INNER_FOLDS, shuffle=True, random_state=seed)

    if IS_BINARY:
        oof_prob = np.zeros(len(y), dtype=np.float64)
    else:
        oof_prob = np.zeros((len(y), N_CLASSES), dtype=np.float64)

    pos_weights_used = []

    for inner_fold, (tr_idx, va_idx) in enumerate(skf.split(X_df, y), start=1):
        X_tr_df = X_df.iloc[tr_idx].reset_index(drop=True)
        X_va_df = X_df.iloc[va_idx].reset_index(drop=True)
        y_tr = y[tr_idx]
        y_va = y[va_idx]

        prep = FoldPreprocessor()
        X_tr = prep.fit_transform(X_tr_df)
        X_va = prep.transform(X_va_df)

        model, best_epoch, best_val_loss, pos_weight_value = train_one_model(
            X_tr, y_tr,
            X_va, y_va,
            hp=hp,
            max_epochs=MAX_EPOCHS_INNER,
            patience=PATIENCE_INNER,
            seed=seed + inner_fold * 77
        )
        pos_weights_used.append(float(pos_weight_value))

        prob = predict_proba(model, X_va, batch_size=max(1024, hp["batch_size"]))
        oof_prob[va_idx] = prob

    if IS_BINARY:
        best_thr = tune_threshold_constrained(
            y_true=y,
            prob=oof_prob,
            grid=THRESHOLD_GRID,
            acc_drop_tol=ACC_DROP_TOL,
            min_dynamic_precision=MIN_DYNAMIC_PRECISION,
            precision_relax=PRECISION_RELAX,
        )
        metrics = compute_metrics(y, oof_prob, threshold=best_thr)
    else:
        best_thr = None
        metrics = compute_metrics(y, oof_prob, threshold=None)

    return {
        "hp": hp,
        "threshold": best_thr,
        "metrics": metrics,
        "score": metrics["accuracy"],
        "mean_pos_weight": float(np.mean(pos_weights_used)) if len(pos_weights_used) > 0 else 1.0,
    }

set_seed(SEED)
ensure_dir(OUTPUT_DIR)

print("\n" + "=" * 80)
print(f"[{now()}] LOADING DATA")
print("=" * 80)

train_df = safe_read_csv(TRAIN_PATH)
test_df = safe_read_csv(TEST_PATH)

if TARGET_COLUMN not in train_df.columns:
    raise ValueError(f"Target column '{TARGET_COLUMN}' not found.")

drop_train = list(set(ID_COLUMNS + DROP_COLUMNS + [TARGET_COLUMN]))
X_train_df = train_df.drop(columns=[c for c in drop_train if c in train_df.columns], errors="ignore").copy()
y_raw = train_df[TARGET_COLUMN].copy()

le = LabelEncoder()
y = le.fit_transform(y_raw)

class_names = [str(x) for x in le.classes_]
N_CLASSES = len(class_names)
IS_BINARY = (N_CLASSES == 2)

print(f"Train shape   : {train_df.shape}")
print(f"Feature shape : {X_train_df.shape}")
print(f"Classes       : {class_names}")
print(f"N_CLASSES     : {N_CLASSES}")
print(f"Task type     : {'binary' if IS_BINARY else 'multiclass'}")
print(f"Device        : {DEVICE}")

if test_df is not None:
    drop_test = list(set(ID_COLUMNS + DROP_COLUMNS + [TARGET_COLUMN]))
    X_test_df = test_df.drop(columns=[c for c in drop_test if c in test_df.columns], errors="ignore").copy()

    if TARGET_COLUMN in test_df.columns:
        y_test = le.transform(test_df[TARGET_COLUMN])
        has_test_labels = True
    else:
        y_test = None
        has_test_labels = False

    print(f"Test shape    : {test_df.shape}")
    print(f"Test labels   : {'present' if has_test_labels else 'absent'}")
else:
    X_test_df = None
    y_test = None
    has_test_labels = False

print("\n" + "=" * 80)
print(f"[{now()}] STARTING NESTED CV ({OUTER_FOLDS} OUTER x {INNER_FOLDS} INNER)")
print("=" * 80)

outer_cv = StratifiedKFold(n_splits=OUTER_FOLDS, shuffle=True, random_state=SEED)

if IS_BINARY:
    oof_prob = np.zeros(len(y), dtype=np.float64)
else:
    oof_prob = np.zeros((len(y), N_CLASSES), dtype=np.float64)

outer_rows = []
outer_thresholds = []
outer_best_hps = []
outer_test_probs = []

candidate_pool = SEARCH_CANDIDATES[:N_TRIALS]

for outer_fold, (dev_idx, hold_idx) in enumerate(outer_cv.split(X_train_df, y), start=1):
    print(f"\n[{now()}] OUTER FOLD {outer_fold}/{OUTER_FOLDS}")

    X_dev_df = X_train_df.iloc[dev_idx].reset_index(drop=True)
    y_dev = y[dev_idx]

    X_hold_df = X_train_df.iloc[hold_idx].reset_index(drop=True)
    y_hold = y[hold_idx]

    fold_infos = []

    for trial_id, hp in enumerate(candidate_pool, start=1):
        info = inner_cv_score(
            X_df=X_dev_df,
            y=y_dev,
            hp=hp,
            seed=SEED + outer_fold * 100 + trial_id
        )
        fold_infos.append(info)

        m = info["metrics"]

        if IS_BINARY:
            print(
                f"  trial {trial_id}/{len(candidate_pool)} | "
                f"acc={m['accuracy']:.5f} | f1={m['f1']:.5f} | rec={m['recall']:.5f} | "
                f"prec={m['precision']:.5f} | mcc={m['mcc']:.5f} | "
                f"auc={m['roc_auc']:.5f} | logloss={m['logloss']:.5f} | "
                f"thr={info['threshold']:.3f} | pos_w={info['mean_pos_weight']:.3f}"
            )
        else:
            print(
                f"  trial {trial_id}/{len(candidate_pool)} | "
                f"acc={m['accuracy']:.5f} | f1={m['f1']:.5f} | rec={m['recall']:.5f} | "
                f"prec={m['precision']:.5f} | mcc={m['mcc']:.5f} | "
                f"auc={m['roc_auc']:.5f} | logloss={m['logloss']:.5f} | "
                f"class_w={info['mean_pos_weight']:.3f}"
            )

    best_info = choose_best_hp_info(fold_infos, acc_tol=HP_ACC_TOL)
    best_hp = best_info["hp"]
    best_thr = best_info["threshold"]

    outer_best_hps.append(best_hp)
    outer_thresholds.append(best_thr)

    print(f"  best hp  : {hp_to_str(best_hp)}")
    if IS_BINARY:
        print(f"  best thr : {best_thr:.3f}")

    X_tr_df, X_va_df, y_tr, y_va = train_test_split(
        X_dev_df, y_dev,
        test_size=OUTER_DEV_VALID_SIZE,
        random_state=SEED + outer_fold,
        stratify=y_dev
    )

    prep = FoldPreprocessor()
    X_tr = prep.fit_transform(X_tr_df)
    X_va = prep.transform(X_va_df)
    X_hold = prep.transform(X_hold_df)

    model, best_epoch, best_val_loss, pos_weight_value = train_one_model(
        X_tr, y_tr,
        X_va, y_va,
        hp=best_hp,
        max_epochs=MAX_EPOCHS_OUTER,
        patience=PATIENCE_OUTER,
        seed=SEED + outer_fold * 1000
    )

    hold_prob = predict_proba(model, X_hold, batch_size=max(1024, best_hp["batch_size"]))
    oof_prob[hold_idx] = hold_prob

    fold_metrics = compute_metrics(y_hold, hold_prob, threshold=best_thr if IS_BINARY else None)
    fold_metrics["outer_fold"] = outer_fold
    fold_metrics["best_epoch"] = best_epoch
    fold_metrics["best_val_loss"] = best_val_loss
    fold_metrics["pos_weight"] = float(pos_weight_value)
    outer_rows.append(fold_metrics)

    print(
        f"  fold result | acc={fold_metrics['accuracy']:.5f} | "
        f"prec={fold_metrics['precision']:.5f} | rec={fold_metrics['recall']:.5f} | "
        f"spec={fold_metrics['specificity']:.5f} | f1={fold_metrics['f1']:.5f} | "
        f"mcc={fold_metrics['mcc']:.5f} | auc={fold_metrics['roc_auc']:.5f} | "
        f"logloss={fold_metrics['logloss']:.5f}"
    )

    if X_test_df is not None:
        X_test_proc = prep.transform(X_test_df)
        test_prob_fold = predict_proba(model, X_test_proc, batch_size=max(1024, best_hp["batch_size"]))
        outer_test_probs.append(test_prob_fold)
outer_df = pd.DataFrame(outer_rows)

if IS_BINARY:
    global_threshold = tune_threshold_constrained(
        y_true=y,
        prob=oof_prob,
        grid=THRESHOLD_GRID,
        acc_drop_tol=ACC_DROP_TOL,
        min_dynamic_precision=MIN_DYNAMIC_PRECISION,
        precision_relax=PRECISION_RELAX,
    )
else:
    global_threshold = None

oof_metrics = compute_metrics(y, oof_prob, threshold=global_threshold if IS_BINARY else None)

print("\n" + "=" * 80)
print(f"[{now()}] NESTED CV SUMMARY")
print("=" * 80)
print(outer_df[
    ["outer_fold", "accuracy", "precision", "recall", "specificity", "f1",
     "mcc", "balanced_accuracy", "roc_auc", "pr_auc", "logloss", "threshold", "pos_weight"]
].round(5).to_string(index=False))

print("\nOOF SUMMARY")
print(pd.DataFrame([oof_metrics]).round(5).to_string(index=False))

if SHOW_PLOTS:
    if IS_BINARY:
        cm_oof = confusion_matrix(y, (oof_prob >= global_threshold).astype(int), labels=[0, 1])
        plot_conf_matrix(
            cm_oof,
            title=f"OOF Confusion Matrix (thr={global_threshold:.3f})",
            class_names=class_names
        )
    else:
        cm_oof = confusion_matrix(y, np.argmax(oof_prob, axis=1), labels=np.arange(N_CLASSES))
        plot_conf_matrix(
            cm_oof,
            title="OOF Confusion Matrix",
            class_names=class_names
        )

hp_keys = [tuple(sorted(hp.items())) for hp in outer_best_hps]
hp_counter = Counter(hp_keys)
max_count = max(hp_counter.values())
candidates = [dict(k) for k, v in hp_counter.items() if v == max_count]

if len(candidates) == 1:
    final_best_hp = candidates[0]
else:
    rows = []
    for cand in candidates:
        mask = [hp == cand for hp in outer_best_hps]
        rows.append({
            "hp": cand,
            "mean_acc": float(outer_df.loc[mask, "accuracy"].mean()),
            "mean_f1": float(outer_df.loc[mask, "f1"].mean()),
            "mean_recall": float(outer_df.loc[mask, "recall"].mean()),
            "mean_mcc": float(outer_df.loc[mask, "mcc"].mean()),
            "mean_logloss": float(outer_df.loc[mask, "logloss"].mean()),
        })

    hp_df = pd.DataFrame(rows)
    best_mean_acc = float(hp_df["mean_acc"].max())

    keep = hp_df.loc[hp_df["mean_acc"] >= best_mean_acc - HP_ACC_TOL].copy()
    keep = keep.sort_values(
        by=["mean_f1", "mean_recall", "mean_mcc", "mean_acc", "mean_logloss"],
        ascending=[False, False, False, False, True]
    ).reset_index(drop=True)

    final_best_hp = keep.iloc[0]["hp"]

print("\n" + "=" * 80)
print(f"[{now()}] FINAL CHOSEN SETTINGS")
print("=" * 80)
print("Final best hyperparameters:")
print(json.dumps(final_best_hp, indent=2))
if IS_BINARY:
    print(f"OOF-derived threshold: {global_threshold:.4f}")

final_models = []
final_preps = []
final_valid_probs = []
final_valid_targets = []

for i, seed_i in enumerate(FINAL_ENSEMBLE_SEEDS, start=1):
    print(f"\n[{now()}] FINAL MODEL {i}/{len(FINAL_ENSEMBLE_SEEDS)}")

    X_tr_df, X_va_df, y_tr, y_va = train_test_split(
        X_train_df, y,
        test_size=FINAL_VALID_SIZE,
        random_state=seed_i,
        stratify=y
    )

    prep = FoldPreprocessor()
    X_tr = prep.fit_transform(X_tr_df)
    X_va = prep.transform(X_va_df)

    model, best_epoch, best_val_loss, pos_weight_value = train_one_model(
        X_tr, y_tr,
        X_va, y_va,
        hp=final_best_hp,
        max_epochs=MAX_EPOCHS_FINAL,
        patience=PATIENCE_FINAL,
        seed=seed_i
    )

    final_models.append(model)
    final_preps.append(prep)

    va_prob = predict_proba(model, X_va, batch_size=max(1024, final_best_hp["batch_size"]))
    final_valid_probs.append(va_prob)
    final_valid_targets.append(np.asarray(y_va, dtype=int))

    print(
        f"  best_epoch={best_epoch} | best_val_loss={best_val_loss:.5f} | "
        f"pos_weight={pos_weight_value:.3f}"
    )

def predict_ensemble(models, preps, X_df, batch_size=1024):
    probs = []
    for model, prep in zip(models, preps):
        X_proc = prep.transform(X_df)
        p = predict_proba(model, X_proc, batch_size=batch_size)
        probs.append(p)
    return np.mean(np.stack(probs, axis=0), axis=0)

if IS_BINARY:
    if len(final_valid_probs) > 0:
        final_valid_prob = np.concatenate(final_valid_probs).astype(np.float64)
        final_valid_y = np.concatenate(final_valid_targets).astype(int)

        final_threshold = tune_threshold_constrained(
            y_true=final_valid_y,
            prob=final_valid_prob,
            grid=THRESHOLD_GRID,
            acc_drop_tol=ACC_DROP_TOL,
            min_dynamic_precision=MIN_DYNAMIC_PRECISION,
            precision_relax=PRECISION_RELAX,
        )
    else:
        final_threshold = global_threshold

    print(f"\n[{now()}] FINAL DEPLOYMENT THRESHOLD: {final_threshold:.4f}")
else:
    final_threshold = None
    print(f"\n[{now()}] FINAL DEPLOYMENT MODE: argmax over class probabilities")

if X_test_df is not None:
    print("\n" + "=" * 80)
    print(f"[{now()}] TEST RESULT")
    print("=" * 80)

    final_test_prob = predict_ensemble(
        final_models,
        final_preps,
        X_test_df,
        batch_size=max(1024, final_best_hp["batch_size"])
    )

    if IS_BINARY:
        final_test_pred = (final_test_prob >= final_threshold).astype(int)
    else:
        final_test_pred = np.argmax(final_test_prob, axis=1)

    if has_test_labels:
        test_metrics = compute_metrics(y_test, final_test_prob, threshold=final_threshold if IS_BINARY else None)
        print(pd.DataFrame([test_metrics]).round(5).to_string(index=False))

        if SHOW_PLOTS:
            if IS_BINARY:
                cm_test = confusion_matrix(y_test, final_test_pred, labels=[0, 1])
                plot_conf_matrix(
                    cm_test,
                    title=f"Test Confusion Matrix (thr={final_threshold:.3f})",
                    class_names=class_names
                )
            else:
                cm_test = confusion_matrix(y_test, final_test_pred, labels=np.arange(N_CLASSES))
                plot_conf_matrix(
                    cm_test,
                    title="Test Confusion Matrix",
                    class_names=class_names
                )

        if len(outer_test_probs) > 0:
            nested_test_prob = np.mean(np.stack(outer_test_probs, axis=0), axis=0)
            nested_test_metrics = compute_metrics(y_test, nested_test_prob, threshold=global_threshold if IS_BINARY else None)

            print("\nOuter-fold test ensemble reference:")
            print(pd.DataFrame([nested_test_metrics]).round(5).to_string(index=False))
    else:
        test_metrics = None
        print("Test labels not available.")
else:
    final_test_prob = None
    final_test_pred = None
    test_metrics = None

if SAVE_OUTPUTS:
    ensure_dir(OUTPUT_DIR)

    outer_df.to_csv(os.path.join(OUTPUT_DIR, "outer_fold_results.csv"), index=False)
    pd.DataFrame([oof_metrics]).to_csv(os.path.join(OUTPUT_DIR, "oof_metrics.csv"), index=False)

    with open(os.path.join(OUTPUT_DIR, "final_best_hyperparameters.json"), "w") as f:
        json.dump(
            {
                "best_hyperparameters": final_best_hp,
                "oof_global_threshold": global_threshold,
                "final_threshold": final_threshold,
                "classes": class_names,
                "n_classes": N_CLASSES,
                "task_type": "binary" if IS_BINARY else "multiclass",
                "device": str(DEVICE),
                "model_name": "TabFGT",
                "threshold_grid_start": float(THRESHOLD_GRID[0]),
                "threshold_grid_end": float(THRESHOLD_GRID[-1]),
                "acc_drop_tol": float(ACC_DROP_TOL),
                "min_dynamic_precision": float(MIN_DYNAMIC_PRECISION),
                "precision_relax": float(PRECISION_RELAX),
                "use_mild_pos_weight": bool(USE_MILD_POS_WEIGHT),
                "pos_weight_power": float(POS_WEIGHT_POWER),
                "pos_weight_max": float(POS_WEIGHT_MAX),
                "hp_acc_tol": float(HP_ACC_TOL),
            },
            f,
            indent=2
        )

    if IS_BINARY:
        oof_pred = (oof_prob >= global_threshold).astype(int)
        oof_df = pd.DataFrame({
            "true_label_encoded": y,
            "true_label": [class_names[i] for i in y],
            "prob_positive": oof_prob,
            "pred_label_encoded": oof_pred,
            "pred_label": [class_names[i] for i in oof_pred],
        })
    else:
        oof_pred = np.argmax(oof_prob, axis=1)
        oof_df = pd.DataFrame({
            "true_label_encoded": y,
            "true_label": [class_names[i] for i in y],
            "pred_label_encoded": oof_pred,
            "pred_label": [class_names[i] for i in oof_pred],
        })
        for i, cls_name in enumerate(class_names):
            oof_df[f"prob_class_{cls_name}"] = oof_prob[:, i]

    if ID_COLUMNS:
        ids = [c for c in ID_COLUMNS if c in train_df.columns]
        if len(ids) > 0:
            oof_df = pd.concat([train_df[ids].reset_index(drop=True), oof_df], axis=1)
    oof_df.to_csv(os.path.join(OUTPUT_DIR, "oof_predictions.csv"), index=False)

    if X_test_df is not None:
        if IS_BINARY:
            test_out = pd.DataFrame({
                "prob_positive": final_test_prob,
                "pred_label_encoded": final_test_pred,
                "pred_label": [class_names[i] for i in final_test_pred],
            })
        else:
            test_out = pd.DataFrame({
                "pred_label_encoded": final_test_pred,
                "pred_label": [class_names[i] for i in final_test_pred],
            })
            for i, cls_name in enumerate(class_names):
                test_out[f"prob_class_{cls_name}"] = final_test_prob[:, i]

        if ID_COLUMNS:
            ids = [c for c in ID_COLUMNS if c in test_df.columns]
            if len(ids) > 0:
                test_out = pd.concat([test_df[ids].reset_index(drop=True), test_out], axis=1)
        test_out.to_csv(os.path.join(OUTPUT_DIR, "test_predictions.csv"), index=False)

    print("\nSaved outputs in:", OUTPUT_DIR)

print("\n" + "=" * 80)
print(f"[{now()}] FINAL REPORT")
print("=" * 80)
print(f"Model         : TabFGT")
print(f"Task type     : {'binary' if IS_BINARY else 'multiclass'}")
print(f"Final HP      : {hp_to_str(final_best_hp)}")

if IS_BINARY:
    print(f"OOF thr       : {global_threshold:.4f}")
    print(f"Final thr     : {final_threshold:.4f}")

print(f"OOF accuracy  : {oof_metrics['accuracy']:.5f}")
print(f"OOF precision : {oof_metrics['precision']:.5f}")
print(f"OOF recall    : {oof_metrics['recall']:.5f}")
print(f"OOF spec      : {oof_metrics['specificity']:.5f}")
print(f"OOF f1        : {oof_metrics['f1']:.5f}")
print(f"OOF mcc       : {oof_metrics['mcc']:.5f}")
print(f"OOF roc_auc   : {oof_metrics['roc_auc']:.5f}")
print(f"OOF logloss   : {oof_metrics['logloss']:.5f}")

if test_metrics is not None:
    print("\nTEST headline metrics")
    print(f"Test accuracy : {test_metrics['accuracy']:.5f}")
    print(f"Test precision: {test_metrics['precision']:.5f}")
    print(f"Test recall   : {test_metrics['recall']:.5f}")
    print(f"Test spec     : {test_metrics['specificity']:.5f}")
    print(f"Test f1       : {test_metrics['f1']:.5f}")
    print(f"Test mcc      : {test_metrics['mcc']:.5f}")
    print(f"Test roc_auc  : {test_metrics['roc_auc']:.5f}")
    print(f"Test logloss  : {test_metrics['logloss']:.5f}")
```
